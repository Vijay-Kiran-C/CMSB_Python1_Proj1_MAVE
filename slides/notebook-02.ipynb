# Session 2 — Questions about one sequence

What can we learn from one DNA sequence using a few existing Python operations?

## 1. Store one sequence

We will use the same short sequence for most of today's questions.

```python
sequence = "ATGCGCTA"
```

## 2. Select a consecutive piece

A slice includes its starting position and stops before its stopping position.

```python
sequence[:3]
```

```python
sequence[3:]
```

```python
sequence[2:6]
```

```python
len(sequence[:3])
```

## 3. Count symbols and calculate GC fraction

The existing string operation `.count(...)` counts occurrences of the exact string supplied in parentheses.

```python
sequence.count("G")
```

```python
sequence.count("C")
```

```python
sequence.count("GC")
```

```python
g_count = sequence.count("G")
c_count = sequence.count("C")
gc_count = g_count + c_count
gc_fraction = gc_count / len(sequence)
```

```python
g_count, c_count, gc_count, gc_fraction
```

## 4. Ask yes-or-no questions

`==` asks whether two values are equal. `in` asks whether one string occurs inside another.

```python
sequence[:3] == "ATG"
```

```python
"GCG" in sequence
```

```python
"AAA" in sequence
```

```python
gc_fraction > 0.5
```

```python
type("GCG" in sequence)
```

## 5. Produce a new string without changing the original

The method returns a new string. The two names below can therefore refer to different values.

```python
sequence = "ATGCGCTA"
rna_sequence = sequence.replace("T", "U")
```

```python
sequence, rna_sequence
```

Replacing T with U here only changes the alphabet used for the same symbol string. It does not by itself model transcription or determine which DNA strand is represented.

### Extension — capitalization

```python
mixed_sequence = "atgcGCTA"
upper_sequence = mixed_sequence.upper()
```

```python
mixed_sequence, upper_sequence
```

```python
mixed_sequence.count("G"), upper_sequence.count("G")
```

## 6. Audit a calculation

The expression runs successfully, but its numerator may not represent the intended quantity.

```python
g_fraction = sequence.count("G") / len(sequence)
g_fraction
```

```python
gc_fraction = (sequence.count("G") + sequence.count("C")) / len(sequence)
gc_fraction
```

## Summary

One string can be selected, counted, compared, and transformed. Each result still needs an interpretation tied to the biological question.
