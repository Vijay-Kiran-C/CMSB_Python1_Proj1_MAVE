# Session 3 — Making decisions

How can a program apply a biological rule?

## 1. Ask a yes-or-no question

The membership expression produces a Boolean value.

```python
sequence = "ATGNCG"
"N" in sequence
```

```python
type("N" in sequence)
```

## 2. Use the answer to choose a branch

An `if`/`else` statement selects one indented block.

```python
sequence = "ATGNCG"

if "N" in sequence:
    status = "review"
else:
    status = "ready"

status
```

```python
sequence = "ATGCCG"

if "N" in sequence:
    status = "review"
else:
    status = "ready"

status
```

## 3. Indentation marks each block

The unindented assignment after the conditional runs after either branch.

```python
sequence = "ATGCCG"

if "N" in sequence:
    message = "ambiguous symbol observed"
    action = "inspect the sequence"
else:
    message = "no N symbol observed"
    action = "continue this analysis"

summary = message + "; " + action
summary
```

## 4. Represent three mutually exclusive outcomes

An `elif` condition is considered only when the earlier `if` condition is false.

```python
gc_fraction = 0.5

if gc_fraction > 0.5:
    gc_label = "above half"
elif gc_fraction == 0.5:
    gc_label = "exactly half"
else:
    gc_label = "below half"

gc_label
```

These labels describe comparison with a chosen threshold. They are not universal biological categories.

## 5. Modify a sequence rule

```python
sequence = "ATGCCG"

if sequence[:3] == "ATG":
    start_label = "begins with ATG"
else:
    start_label = "does not begin with ATG"

start_label
```

## 6. Audit the claim

The condition below checks only for uppercase `N`. Its labels make a stronger claim than that check supports.

```python
sequence = "ATGCCG"

if "N" in sequence:
    status = "invalid sequence"
else:
    status = "valid sequence"

status
```

`N` commonly represents an unresolved nucleotide identity. Whether it is acceptable depends on the analysis, and absence of `N` is not complete validation.

## Summary

A conditional connects a Boolean question to a selected action. The program can apply its stated rule correctly even when a biological label overstates what was checked.
