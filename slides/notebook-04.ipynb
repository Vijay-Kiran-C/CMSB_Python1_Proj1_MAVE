# Session 4 — Repeating an operation

How can a program inspect every symbol in a sequence?

## 1. Visit each symbol

A `for` loop gives the loop variable the next value from the sequence, then runs the indented body.

```python
sequence = "ATGC"

for base in sequence:
    print(base)
```

## 2. Evaluate an expression each iteration

```python
sequence = "ATGC"

for base in sequence:
    is_g = base == "G"
    print(base, is_g)
```

## 3. Make one decision per symbol

The loop controls repetition. The conditional chooses an action during the current iteration.

```python
sequence = "ATNG"

for base in sequence:
    if base == "N":
        print("ambiguous")
    else:
        print("observed", base)
```

`N` is treated here as a symbol to notice. Whether it is acceptable depends on the analysis.

## 4. Compare statements inside and outside the loop body

```python
for base in "ATGC":
    print(base)
print("done")
```

```python
for base in "ATGC":
    print(base)
    print("done")
```

## 5. Modify the input and repeated question

```python
sequence = "GGTA"

for base in sequence:
    is_g = base == "G"
    print(base, is_g)
```

## 6. Audit the repeated operation

This loop repeats, but the body refers to the whole sequence rather than the current symbol.

```python
sequence = "ATGC"

for base in sequence:
    print(sequence)
```

## Summary

A `for` loop visits values in order and repeats an indented body. Correct repetition still depends on using the intended current value inside that body.
