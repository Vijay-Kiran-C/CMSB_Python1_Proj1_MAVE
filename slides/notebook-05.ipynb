# Session 5 — Counting and constructing

How can repeated steps produce one quantitative summary?

## 1. Keep a count across iterations

An accumulator is initialized before the loop and updated from its current value during selected iterations.

```python
sequence = "GATC"
gc_count = 0

for base in sequence:
    if base in "GC":
        gc_count = gc_count + 1

print(gc_count)
```

## 2. Inspect intermediate state

Temporary diagnostic output can show the current base, condition, and count after every iteration.

```python
sequence = "GATC"
gc_count = 0

for base in sequence:
    is_gc = base in "GC"
    if is_gc:
        gc_count = gc_count + 1
    print(base, is_gc, gc_count)
```

## 3. Diagnose misplaced initialization

This version runs, but it resets the accumulator during every iteration.

```python
sequence = "GATC"

for base in sequence:
    gc_count = 0
    if base in "GC":
        gc_count = gc_count + 1

print(gc_count)
```

## 4. Diagnose an update on the wrong iterations

Indentation determines whether the update belongs to the conditional block.

```python
sequence = "GATC"
gc_count = 0

for base in sequence:
    if base in "GC":
        print("GC symbol observed")
    gc_count = gc_count + 1

print(gc_count)
```

## 5. Construct a complementary string

A string accumulator begins with the empty string and gains one complementary base per iteration.

```python
sequence = "ATGC"
complement = ""

for base in sequence:
    if base == "A":
        complement_base = "T"
    elif base == "T":
        complement_base = "A"
    elif base == "G":
        complement_base = "C"
    else:
        complement_base = "G"

    complement = complement + complement_base

print(complement)
```

## 6. Audit the biological result

For `5'-ATGC-3'`, this loop constructs `TACG`. It does not reverse that text into the usual 5'-to-3' reverse complement, `GCAT`. The final `else` also assumes that any other symbol should be treated as C, so the calculation does not safely handle ambiguity symbols such as N.

## Summary

An accumulator has an initial value, persists across iterations, and is updated from its current state. Intermediate state can localize a defect that a final result alone cannot explain.
