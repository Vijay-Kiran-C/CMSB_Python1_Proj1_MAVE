# Session 6 — Functions as testable claims

How can we turn a define a reliable, repeated operation?

## 1. Begin with a contract

`gc_fraction` accepts one non-empty string containing only uppercase A, C, G, and T. It returns a float equal to the number of G and C symbols divided by the total number of symbols. It does not change the input string.

## 2. Define the function

Running a `def` statement creates the function. A later call supplies an argument and executes the body.

```python
def gc_fraction(sequence):
    gc_count = 0
    for base in sequence:
        if base in "GC":
            gc_count = gc_count + 1
    return gc_count / len(sequence)
```

## 3. Call the function and use its returned value

```python
result = gc_fraction("GATC")
print(result)
```

```python
first = gc_fraction("GGCC")
second = gc_fraction("AATT")
difference = first - second
print(difference)
```

## 4. Check known-answer examples

An assertion records an expected result and reports a failure when the observed result differs.

```python
assert gc_fraction("GGCC") == 1.0
assert gc_fraction("AATT") == 0.0
assert gc_fraction("ATGC") == 0.5
```

## 5. Test a plausible error

This version accidentally counts G but not C. A C-only input distinguishes it directly from the intended calculation.

```python
def gc_fraction_buggy(sequence):
    gc_count = 0
    for base in sequence:
        if base == "G":
            gc_count = gc_count + 1
    return gc_count / len(sequence)
```

```python
print(gc_fraction("CCCC"))
print(gc_fraction_buggy("CCCC"))
```

## 6. Audit the input domain

Lowercase, ambiguity symbols, and empty input are outside the first contract. Producing a number does not make an outside-contract result biologically valid; the promise and implementation must be revised together if those inputs should be supported.

```python
print(gc_fraction("atgc"))
print(gc_fraction("ATNC"))
```

## Summary

A trustworthy function joins a clear contract, traceable implementation, returned value, and tests chosen to expose plausible errors. Passing tests verifies specific expectations.
