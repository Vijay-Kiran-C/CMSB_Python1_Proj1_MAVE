## 3. Inspect positions in the sequence

Python numbers positions from zero. A negative position counts back from the end.

| Python position | 0 | 1 | 2 | 3 | 4 | 5 |
|---|---:|---:|---:|---:|---:|---:|
| Base | A | T | G | C | G | T |
| Position from end | -6 | -5 | -4 | -3 | -2 | -1 |

```python
sequence[0]
```

```python
sequence[3]
```

```python
sequence[-1]
```

```python
type(sequence[0])
```
